# Mathplotlib

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

```python
plt.subplot(2,2,4)
plt.plot(x**4, y)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.3,0.4,0.5])
axis2.plot(x,y)
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
plt.savefig("BoxAndWhiskerPlot.png")
```
