```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,6,])
y = np.linspace(0,14,5)
```

```python
x
```

```python
y
```

```python
plt.plot(x,y)
plt.title("Hamara Marzi")
plt.xlabel("Khalaga")
plt.ylabel("Nahi Khalage")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y**2)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(x**2,x**2)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
fig = plt.figure(figsize = (20,20),dpi = 200)
fig = plt.figure()
axis1 = fig.add_axes([1,1,1,1])
axis1.plot(x,y**2)
axis2 = fig.add_axes([0.30,0.30,.6,.6])
axis2.plot(x**2,y)
```
