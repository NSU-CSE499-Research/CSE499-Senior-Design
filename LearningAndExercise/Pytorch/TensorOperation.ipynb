1 student calculate CGPA and Placement

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0,requires_grad=True)
b = torch.tensor(0.0,requires_grad=True)
```

```python
z = w*x+b
```

```python
z
```

```python
Y_Pred = torch.sigmoid(z)
```

```python
y
```

```python
import torch.nn as nn

binary_cross_entropy_loss = nn.BCELoss()
```

```python
loss = binary_cross_entropy_loss(
    Y_Pred,
    y.float().view_as(Y_Pred)
)
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```
