```python
def Extract_Episodes(text):
    check = False
    Epi = ''
    for i in text:
        if(i == '('):
            check = True
        elif(i == ')'):
            check = False
            break
        elif(check == True):
            Epi+=i
    return Epi
```

```python
l={}
for i in range(0,5,):
    l[i]=Extract_Episodes(df.loc[i]['Title'])
    
```

```python
df['Episodes']= df['Title'].apply(Extract_Episodes)
```

```python
def TimeStamp(title):
    T = False
    Time = ''
    for i in title:
        if i ==')':
            T = True
            for j in range(i,18,1):
                Time[j]+=title[i]
                i = i+1
            T = True
    return Time
```
