```python
%cd /content/LLaVA

import torch

from llava.model.builder import load_pretrained_model
from llava.mm_utils import get_model_name_from_path

model_path = "liuhaotian/llava-v1.5-7b"
model_name = get_model_name_from_path(model_path)

tokenizer, model, image_processor, context_len = load_pretrained_model(
    model_path=model_path,
    model_base=None,
    model_name=model_name,
    load_4bit=True,
    device_map="auto"
)

model.eval()

print("Model loaded successfully")
print("Model:", model_name)
print("Context length:", context_len)
```

```python
import os

from google.colab import files
from PIL import Image
from IPython.display import display

uploaded = files.upload()

image_path = os.path.abspath(next(iter(uploaded.keys())))
image = Image.open(image_path).convert("RGB")

display(image)
print("Image path:", image_path)
```

```python
from llava.constants import IMAGE_TOKEN_INDEX, DEFAULT_IMAGE_TOKEN
from llava.conversation import conv_templates
from llava.mm_utils import process_images, tokenizer_image_token

def ask_llava(image_path, question):
    image = Image.open(image_path).convert("RGB")
    image_size = image.size

    image_tensor = process_images(
        [image],
        image_processor,
        model.config
    ).to(model.device, dtype=torch.float16)

    prompt_question = DEFAULT_IMAGE_TOKEN + "\n" + question

    conversation = conv_templates["llava_v1"].copy()
    conversation.append_message(
        conversation.roles[0],
        prompt_question
    )
    conversation.append_message(
        conversation.roles[1],
        None
    )

    prompt = conversation.get_prompt()

    input_ids = tokenizer_image_token(
        prompt,
        tokenizer,
        IMAGE_TOKEN_INDEX,
        return_tensors="pt"
    ).unsqueeze(0).to(model.device)

    with torch.inference_mode():
        output_ids = model.generate(
            input_ids,
            images=image_tensor,
            image_sizes=[image_size],
            do_sample=False,
            max_new_tokens=256,
            use_cache=True
        )

    answer = tokenizer.batch_decode(
        output_ids,
        skip_special_tokens=True
    )[0].strip()

    return answer
```

```python
question = "Describe this image in detail."

answer = ask_llava(image_path, question)

print("Question:", question)
print("LLaVA:", answer)
```
